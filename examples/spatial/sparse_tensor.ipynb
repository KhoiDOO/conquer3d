```python
import sys
import os
if os.path.abspath('..') not in sys.path:
    sys.path.append(os.path.abspath('..'))
import render
import nvdiffrast.torch as dr
import torch
import torch.nn as nn
import torchsparse
import torchsparse.nn as spnn
from torchsparse import SparseTensor
import open3d as o3d
import conquer3d as c3d
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import trimesh
import numpy as np
import kaolin as kal
```

```python
# 1. Construct TriangleMesh from Stanford Bunny
bunny = o3d.data.BunnyMesh()
mesh_o3d = o3d.io.read_triangle_mesh(bunny.path)

verts = torch.tensor(np.asarray(mesh_o3d.vertices), dtype=torch.float32, device="cuda")
tris = torch.tensor(np.asarray(mesh_o3d.triangles), dtype=torch.int32, device="cuda")

verts = verts - verts.mean(0)
verts = verts / verts.norm(dim=-1).max()
verts = verts * 0.8
angle = torch.tensor(np.pi / 2, device="cuda")
R_x = torch.tensor([[1, 0, 0], [0, torch.cos(angle), -torch.sin(angle)], [0, torch.sin(angle), torch.cos(angle)]], device="cuda", dtype=torch.float32)
verts = verts @ R_x.T
mesh = c3d.data_structure.TriangleMesh(verts, tris)
print("TriangleMesh created with", verts.shape[0], "vertices and", tris.shape[0], "triangles.")
```

```python
# 2. Construct Voxel Grid and SDF
res = [64, 64, 64]
grid_vertices, voxels, idx_grids = c3d.data_structure.create_voxel_grid([-1.2, -1.2, -1.2], [1.2, 1.2, 1.2], res)

_, _, _, sdf_dense = mesh.query_points(grid_vertices, return_sdf=True, sign_mode=1)
active_indices = c3d.data_structure.compute_active_voxels(voxels, sdf_dense)
print(f"Active Voxels: {active_indices.shape[0]} / {voxels.shape[0]}")
```

```python
# 3. Create SparseTensor (TorchSparse API)
sparse_coords, sparse_sdfs = c3d.data_structure.voxel2sparse(active_indices, voxels, idx_grids, sdf=sdf_dense)

x = SparseTensor(coords=sparse_coords, feats=sparse_sdfs)
x.stride = (1, 1, 1)
print(f"SparseTensor Input: coords {x.coords.shape}, feats {x.feats.shape}")
```

```python
# 4. Define Improved Simple AutoEncoder
from torchsparse.backbones.modules.blocks import SparseResBlock
class SparseAutoEncoder(nn.Module):
    def __init__(self):
        super().__init__()
        self.encoder = nn.Sequential(
            spnn.Conv3d(8, 32, kernel_size=3, stride=1),
            spnn.BatchNorm(32),
            spnn.ReLU(True),
            SparseResBlock(32, 32, 3),
            SparseResBlock(32, 32, 3),
            spnn.Conv3d(32, 4, kernel_size=3, stride=1)
        )
        self.decoder = nn.Sequential(
            spnn.Conv3d(4, 32, kernel_size=3, stride=1),
            spnn.BatchNorm(32),
            spnn.ReLU(True),
            SparseResBlock(32, 32, 3),
            SparseResBlock(32, 32, 3),
            spnn.Conv3d(32, 8, kernel_size=3, stride=1)
        )
    def forward(self, x):
        z = self.encoder(x)
        out = self.decoder(z)
        return z, out

model = SparseAutoEncoder().cuda()
```

```python
# 5. Train on the Single SparseTensor
optimizer = torch.optim.Adam(model.parameters(), lr=0.005)
criterion = nn.MSELoss()
glctx = dr.RasterizeCudaContext(device="cuda")
gt_kal_mesh = kal.rep.SurfaceMesh(vertices=verts, faces=tris)
train_res = [256, 256]

model.train()
for epoch in range(1000):
    optimizer.zero_grad()
    latent, output = model(x)
    mse_loss = criterion(output.feats, x.feats)
    
    # Differentiable Rendering Loss
    unique_vertices, local_voxels, merged_sdfs = c3d.data_structure.sparse2mesh_topology(
        sparse_coords, output.feats, grid_min=[-1.2, -1.2, -1.2], grid_max=[1.2, 1.2, 1.2], res=res
    )
    mc_vertices, mc_faces, _, _ = c3d.ops.diff_marching_cubes(unique_vertices, local_voxels, merged_sdfs, iso=0.0)
    
    render_loss = torch.tensor(0.0, device='cuda')
    if mc_vertices.shape[0] > 0:
        cameras = render.get_random_camera_batch(4, iter_res=train_res, device="cuda")
        target = render.render_mesh(glctx, gt_kal_mesh, cameras, train_res, return_types=["mask", "depth"])
        extracted_mesh = kal.rep.SurfaceMesh(vertices=mc_vertices, faces=mc_faces.int())
        buffers = render.render_mesh(glctx, extracted_mesh, cameras, train_res, return_types=["mask", "depth"])
        mask_loss = (buffers['mask'] - target['mask']).abs().mean()
        depth_loss = (((((buffers['depth'] - target['depth']) * target['mask'])**2).sum(-1)+1e-8)).sqrt().mean() * 10
        render_loss = mask_loss + depth_loss

    loss = mse_loss + render_loss * 0.01
    loss.backward()
    optimizer.step()
    if (epoch+1) % 100 == 0:
        print(f"Epoch {epoch+1}, MSE Loss: {mse_loss.item():.6f}, Render Loss: {render_loss.item():.6f}")

model.eval()
with torch.no_grad():
    latent, output = model(x)
print("Training complete.")
```

```python
# 6. Perform Marching Cubes on Input, Latent, Output
def extract_mesh_from_feats(feats_n8):
    unique_vertices, local_voxels, merged_sdfs = c3d.data_structure.sparse2mesh_topology(
        sparse_coords, feats_n8, grid_min=[-1.2, -1.2, -1.2], grid_max=[1.2, 1.2, 1.2], res=res
    )
    vert, tri, _, _ = c3d.ops.diff_marching_cubes(unique_vertices, local_voxels, merged_sdfs, iso=0.0)
    return vert, tri

vert_in, tri_in = extract_mesh_from_feats(x.feats)
latent_feats_8 = latent.feats.repeat(1, 2)
vert_z, tri_z = extract_mesh_from_feats(latent_feats_8)
vert_out, tri_out = extract_mesh_from_feats(output.feats)

print(f"Input Mesh: {vert_in.shape[0]} verts, {tri_in.shape[0]} tris")
print(f"Latent Mesh: {vert_z.shape[0]} verts, {tri_z.shape[0]} tris")
print(f"Output Mesh: {vert_out.shape[0]} verts, {tri_out.shape[0]} tris")
```

```python
# 7. Visualize side-by-side
def create_wireframe_trace(vert, tri, name, color):
    if vert.shape[0] == 0 or tri.shape[0] == 0:
        return go.Scatter3d(x=[], y=[], z=[], mode="lines", name=name)
        
    v = vert.cpu().numpy()
    f = tri.cpu().numpy()
    tm = trimesh.Trimesh(vertices=v, faces=f)
    
    x_lines, y_lines, z_lines = [], [], []
    for edge in tm.edges:
        x_lines.extend([v[edge[0], 0], v[edge[1], 0], None])
        y_lines.extend([v[edge[0], 1], v[edge[1], 1], None])
        z_lines.extend([v[edge[0], 2], v[edge[1], 2], None])
        
    return go.Scatter3d(
        x=x_lines, y=y_lines, z=z_lines,
        mode="lines", line=dict(color=color, width=2), name=name
    )

fig = make_subplots(
    rows=1, cols=3, 
    specs=[[{"type": "surface"}, {"type": "surface"}, {"type": "surface"}]],
    subplot_titles=("Input", "Latent", "Output")
)

fig.add_trace(create_wireframe_trace(vert_in, tri_in, "Input", "blue"), row=1, col=1)
fig.add_trace(create_wireframe_trace(vert_z, tri_z, "Latent", "red"), row=1, col=2)
fig.add_trace(create_wireframe_trace(vert_out, tri_out, "Output", "green"), row=1, col=3)

fig.update_layout(height=500, width=1200, showlegend=False)
fig.show()
```
